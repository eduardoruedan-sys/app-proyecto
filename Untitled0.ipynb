```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots()

x = np.linspace(0,10,500)

y = np.piecewise(x,
                 [x<2,
                  (x>=2)&(x<3),
                  (x>=3)&(x<3.2),
                  (x>=3.2)&(x<3.4),
                  (x>=3.4)&(x<5),
                  x>=5],
                 [0,
                  lambda x: 2*(x-2),
                  lambda x: -8*(x-3)+2,
                  lambda x: 6*(x-3.2)-2,
                  0,
                  lambda x: 0.3*np.sin(3*x)])

line, = ax.plot([],[],color="fuchsia",linewidth=3)

ax.set_xlim(0,10)
ax.set_ylim(-3,4)
ax.set_facecolor("black")

def update(frame):
    line.set_data(x[:frame],y[:frame])
    return line,

ani = FuncAnimation(fig,update,frames=len(x),interval=20)

HTML(ani.to_jshtml())
```
